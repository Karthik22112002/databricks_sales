```python
%sql
-- 1. Grant permissions to use the catalog and schema
GRANT USE CATALOG ON CATALOG workspace TO `github-cicd-sp`;
GRANT USE SCHEMA ON SCHEMA workspace.ibm TO `github-cicd-sp`;

-- 2. Grant permissions to read/write the volume
GRANT READ VOLUME ON VOLUME workspace.ibm.capstone_volume TO `github-cicd-sp`;

-- 3. Ensure the service principal can write/modify tables in the schema
GRANT MODIFY, SELECT ON SCHEMA workspace.ibm TO `github-cicd-sp`;
GRANT CREATE TABLE ON SCHEMA workspace.ibm TO `github-cicd-sp`;
```

```python
source_path = "/Volumes/workspace/ibm/capstone_volume/sales_source_1500.csv"

df = (
    spark.read
    .format("csv")
    .option("header", "true")
    .option("inferSchema", "false")
    .load(source_path)
)

df.write \
    .mode("overwrite") \
    .saveAsTable("workspace.ibm.capstone_bronze_sales1")
```
