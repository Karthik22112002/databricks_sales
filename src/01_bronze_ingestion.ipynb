```python
from pyspark.sql.functions import *

source_path = "/Workspace/Users/alihi122007@gmail.com/databricks_sales_capstone/data/sales_source_1500.csv"

df = (
    spark.read
    .format("csv")
    .option("header", "true")
    .option("inferSchema", "false")
    .load(source_path)
)

df.write \
    .mode("overwrite") \
    .saveAsTable("workspace.ibm.capstone_bronze_sales1")

print("Bronze ingestion completed")
print(f"Bronze record count: {df.count()}")

display(df)
df.printSchema()
```
